# Aula 5

Explorando o format
- :s - Texto(strings)
- :d - Inteiros(int)
- :f - Número de ponto flutuante (float)
- :(Número)f - Quantidade de casas decimais (float)
- :(caractere)(>ou< ou ^)(Quantidade)(Tipo - s, d ou f)
- \> - Esquerda
- < - Direita
- ^ - Centro

```python
num1=10
num2=3
div=num1/num2
print("10 dividido por 3 é igual a {:.2f}".format(div))
```

```python
num1=10
num2=3
div=num1/num2
div0='{:.2f}'.format(div)
div1= '{:#>8s}'.format(div0)
div2 = '{:#<12s}'.format(div1)
div_formatado ='{:s}'.format(div2)
print("10 dividido por 3 é igual a", div_formatado)
```
