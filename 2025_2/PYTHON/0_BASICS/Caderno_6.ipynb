# Aula 27
Levantando Exceções em Python (raise)

```python
def divide(n1,n2):
    try:
        return n1/n2
    except ZeroDivisionError as error:
        print(error)
        return False

try:
    print(divide(2,0))
except:
    print('erro') ### O programa não chega aqui, o programa da função já modifica a leitura do python desse try
```

Mas você pode também levantar um erro na função e também para o uso da função

```python
def divide(n1,n2):
    try:
        return n1/n2
    except ZeroDivisionError as error:
        print('Log:',error)
        raise

try:
    print(divide(2,0))
except ZeroDivisionError as error:
    print(error)
```

```python
def divide(n1,n2):
    if n2 == 0:
        raise ValueError ("n2 não pode ser 0")
    return n1/n2

try:
    print(divide(2,0))
except ValueError as error:
    print('Você está tentando dividir por 0')
```

# Aula 29
Como Criar Módulos em Python

Modularizar para organizar a programação

```python
def dobra_lista(lista):
    return [x*2 for x in lista]

def multiplica_lista(lista):
    r=1
    for i in lista:
        r *= i
    return r


lista = [1,2,3,4,5,6]
print(dobra_lista(lista))
print(multiplica_lista(lista))
```

renomeia um arquivo com "o nome da sua biblioteca".py  

e quando for ecrever um código basta importar da forma:
import "o nme da sua biblioteca"

E quando você importa o seu arquivo em outro, você verá que o nome (__name__) será main apenas do arquivo onde está importado.

```python
def dobra_lista(lista):
    return [x*2 for x in lista]

def multiplica_lista(lista):
    r=1
    for i in lista:
        r *= i
    return r

if __name__ == '__main__':                     #só roda no arquivo original, sem ser onde ele foi importado
    lista = [1,2,3,4,5,6]
    print(dobra_lista(lista))
    print(multiplica_lista(lista))
```
