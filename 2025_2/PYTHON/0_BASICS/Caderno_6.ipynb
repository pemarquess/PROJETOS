```python
try:
    print(a)
except NameError as erro:
    print('Erro do desenvolvedor, cobre ele.')
except Exception as erro:
    print('ocorreu um erro inesperado.')

print('Bora continuar')
```

todas mensagens do python de erro terminam com Error, então dá pra escrever Error e procurar quais podem ocorrer

```python
try:
    a=[]
    print(a)
except NameError as erro:
    print('Erro do desenvolvedor, cobre ele.')
except Exception as erro:
    print('ocorreu um erro inesperado.')
else:                                            ### Roda se não ocorrer nenhum erro
    print('Seu código rodou suave')
print('Bora continuar')
```
