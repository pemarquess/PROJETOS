```python
#jogo de forca
segredo = "boyahhh"
letras_tentadas = []
letras_corretas = []
vidas = 5
while vidas > 0:
    letra = input("Digite uma letra:")
    if len(letra)>1:
        print("Só vale uma por vez seu Pilantra")
        continue
    letras_tentadas.append(letra)

    if letra in segredo:
        print(f'Acertou miséra, a letra {letra} existe na palavra secreta.')
        letras_corretas.append(letra)
    else:
        vidas-=1
        print(f'Tú é ruim em, a letra {letra} NÃO EXISTE na palavra secreta. Perdeu uma vida, agora vc tem {vidas}')
    
    segredo_temporario = ''
    for letra_secreta in segredo:
        if letra_secreta in letras_corretas:
            segredo_temporario += letra_secreta
        else: 
            segredo_temporario += '#'

    if segredo_temporario == segredo:
        print(f'BOAAA GANHOU! A palavra era "{segredo}"')
        break
    else:
        print(segredo_temporario)

if vidas==0:
    print("PERDEU!")
```
