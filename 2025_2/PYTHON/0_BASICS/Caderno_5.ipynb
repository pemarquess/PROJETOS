# Aula 21
Sistema de Perguntas e Respostas com Dicionários em Python

```python
perguntas = {
    'Pergunta 1': {
        'pergunta': 'Quanto é 2+2?',
        'respostas': { 'a': '1', 'b': '2', 'c': '3', 'd': '4', 'e': '5'},
        'resposta_certa': 'd'
        },
    'Pergunta 2': {
        'pergunta': 'Quanto é 1x2?',
        'respostas': { 'a': '1', 'b': '2', 'c': '3', 'd': '4', 'e': '5'},
        'resposta_certa': 'b'
    }
}


respostas_certas=0

for pk, pv in perguntas.items():
    print(f'{pk}: {pv["pergunta"]}')

    print('Respostas:')
    for rk, rv in pv['respostas'].items():
        print(f'[{rk}]: {rv}')
    
    resposta_user = input('Sua resposta:')

    if len(resposta_user) > 1:
        print('insira apenas o item. Exemplo: Sua resposta: a')
    else:
        if resposta_user == pv[ 'resposta_certa']:
            print('BOuaaaaaa, acertou!')
            respostas_certas += 1
        else:
            print('Ixi, errou rude!')

    
    print()

qtd_perguntas = len(perguntas)
porcentagem_acerto = respostas_certas/qtd_perguntas * 100

print(f'Você acertou {respostas_certas} respostas')
print(f'Sua porcentagem de acerto foi de {porcentagem_acerto}%')
```
