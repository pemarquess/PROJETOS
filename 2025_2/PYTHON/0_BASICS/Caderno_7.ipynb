Como encontrar arquivos no sistema com Python

```python
import os

caminho_procura = r'C:\URANUS\DOCS'                               ## usar o r no começo para o caso de windows(por causa da barra invertida)
#caminho_procura = '/URANUS/DOCS'    Da certo tb, mas tem o trampo de arrumar as barras
termo_procura = "leibfreid"

for raiz, diretorios, arquivos in os.walk(caminho_procura):       ## Faz o programa acessar todos os arquivos dentro do caminho dado
    for arquivo in arquivos:
        caminho_completo = os.path.join(raiz,arquivo)             ## caminho do C: até o arquivo
        print(caminho_completo)
        nome_arquivo, ext_arquivo = os.path.splitext(arquivo)     ## Separa extensão e nome do arquivo
        print(nome_arquivo,ext_arquivo, sep =' __#__ ')
        tamanho = os.path.getsize(caminho_completo)               ## Dado um arquivo(dado pelo caminho de acesso) ele retorna o tamanho
        print (f'{tamanho} bytes')
```

# Aula 32
Como Mover, Copiar e Apagar Arquivos com Python

```python
import os
import shutil


caminho_original= r'C:\URANUS\PDFS'
caminho_novo= r'C:\URANUS\DOCS'
termos_procura = ['.pdf']

try: 
    os.mkdir(caminho_novo) ## Aqui cria um caminho novo, o try é pq não dá pra criar se o caminho já existe
except FileExistsError as e:
    print(f'Pasta {caminho_novo} já existe.')




for root, dirs, files in os.walk(caminho_original):
    for file in files:
        for termo_procura in termos_procura:
            if termo_procura in file:
                old_file_path = os.path.join(root, file)
                new_file_path = os.path.join(caminho_novo, file)
                name_file, ext_file = os.path.splitext(file)
                shutil.move(old_file_path, new_file_path)
                        # aqui se fizer .copy = copia; .remove = apaga


                print (f'Encontrei o "{name_file}" com extensão "{ext_file}" e movi ele para a pasta {new_file_path}.')
                
```

## O melhor modo de trabalhar em phyton com arquivos (usando gerenciador de contexto):

```python
with open('Caderno7_Out.txt', 'w+') as file:  # é mais ou menos = file=open(...) 
    file.write('Linha 1\n')
    file.write('Linha 2\n')
    file.write('Linha 3\n')
    
    file.seek(0)
    print(file.read())
```
