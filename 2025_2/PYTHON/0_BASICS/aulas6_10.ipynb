# Aula 7
While - Estrutura de Repetição em Python

```python
x=0
while x<2:
    nome = input ("Qual o seu nome??")
    print(f"Olá {nome}")
    x=x+1
print("Quanta gente!!")
```

```python
x=0
while x<10:
    if x==4:
        x=x+1
        continue
    print(x)
    x=x+1
print("cabô")
```

```python
x=0
while x<10:
    if x==4:
        x=x+1
        break
    print(x)
    x=x+1
print("cabô")
```

```python
x=0
while x<5:
    y=0
    while y<3:
        print(f'X é igual a {x} e Y igual a {y}')
        y+=1
    x+=1
print("cabô")
```

```python
#soma de 0 até N
N=3
x=0
y=0

if N%2==0:
    n = N/2
    y = y+n

while x< N/2:
    y=y+N
    x+=1 # x = x + 1
    print(f"tentativa{x} resultado {y}")

print(y)
```

# Aula 8
While/Else - Repetição com Acumuladores em Python

```python
contador=1
acumulador=0
print (contador, acumulador)

while contador <=10:
    acumulador = acumulador + contador
    print (contador, acumulador)
    contador += 1
else:
    print(f"A soma de 0 até 10 é igual a {acumulador}")
```

# Aula 9
Iterando strings com while em Python

```python
string = "~O Rato Roeu a Roupa do Rei de Roma~"
nova_string = ""
c=0

while c< len(string):

    if string[c] == "R":
        nova_string = nova_string + string[c].lower()
    else:
        nova_string = nova_string + string[c]
    c+=1
    print(nova_string)
```

```python
print(nova_string.count('~'))
```

```python
c = 0
contagem_atual = 0
letra = ""

while c< len(string):
    cada_letra = string.count(string[c].strip()) #.strip() remove os espaços

    if contagem_atual < cada_letra and string[c].strip():
        letra = string[c]
        contagem_atual = cada_letra
    c+=1
print(letra,contagem_atual)
```

```python
c = 0
contagem_atual = 0
letra = ""

while c< len(nova_string):
    cada_letra = nova_string.count(nova_string[c].strip()) #.strip() remove os espaços

    if contagem_atual < cada_letra and nova_string[c].strip():
        letra = nova_string[c]
        contagem_atual = cada_letra
    c+=1
print(letra,contagem_atual)
```

# Aula 10
For in - Estrutura de Repetição em Python

```python
texto = "Python"
for letra in texto:
    print (letra)
```

```python
texto = "Python"
for n,letra in enumerate(texto):
    print (n,letra)
```

```python
for número in range(10):
    print(número)
```

```python
for número in range(6,10):
    print(número)
```

```python
for número in range(15,10,-1):
    print(número)
```

```python
for número in range(0,100,17):
    print(número)
```

```python
#soma de 0 a N
N=10
y=0
i=0
for i in range(N+1):
    y=y+i
    i+=1
    print(y)
print(y)
```

```python
texto = "Python"
texto_novo=""

for letra in texto:
    if letra == 't':
        texto_novo = texto_novo + letra.upper()
    elif letra == 'h':
        texto_novo += letra.upper()
    else:
        texto_novo += letra
print(texto_novo)
```
