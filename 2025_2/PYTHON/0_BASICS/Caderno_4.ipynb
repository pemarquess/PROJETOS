## Parte 3 
Funções (def) em Python - args e kwargs

```python
def func(x, y, z, a1=None, a2=None, a3=None):
    print(x, y, z, a1, a2, a3)

func(1,2,3)
func(1,2,3,a3=746)
func(1,2,3,a2=7456)
func(1,2,3,a1=7534)
func(1,2,3,4,5,6)
```

```python
lista = [1,2,3,4,5]
n1, *n2, n3 = lista     # Empacotamento usando *
print(lista)
print(*lista)           # Desempacotamento usando *
print(n1, n2, n3)
```

```python
def func(*args):          #args, nome convencional para argumentos não nomeados
    for v in args:
        print(v)

func(1,2,3,4,5)
```

```python
def func(*args):
    print(args)

lista = [1,2,3,4,5]
func(*lista, 10, 9, 8, 7 ,6)
```

```python
def func(*args, **kwargs):          #kwargs, nome convencional para argumentos nomeados
    print(args)


lista = [1,2,3,4,5]
lista2 = [10, 9, 8, 7, 6]
func(*lista, *lista2, Nome= "Pedro", Sobrenome= "Marques Pontes")
```

```python

def func(*args, **kwargs):          #kwargs, nome convencional para argumentos nomeados
    print(args)
    print(kwargs)

lista = [1,2,3,4,5]
lista2 = [10, 9, 8, 7, 6]
func(*lista, *lista2, Nome= "Pedro", Sobrenome= "Marques Pontes")
```

## Parte 4
Escopo de variáveis de funções em Python - Global e Local

```python
variavel = 'valor'             # Variável de Escopo Global

def func():
    variavel= "valor novo"     # Variável de Escopo Local
    print(variavel)

print(variavel)
func()
print(variavel)
```

```python
variavel = 'valor'             # Variável de Escopo Global

def func():
    global variavel            # "Globaliza" a Variável
    variavel= "valor novo"     # Variável de Escopo Global
    print(variavel)

print(variavel)
func()
print(variavel)
```

# Aula 17
Expressões Lambda(funções anônimas) em Python

```python
a = lambda x, y: x * y
print(a(3,4))
```

```python
Lista0=[
    ['P1', 13],
    ['P2', 52],
    ['P3', 45],
    ['P4', 76],
    ['P5', 21]
]

def func(item):
    return item[1]

Lista=Lista0
Lista.sort(key = func, reverse= True)     
#Key é o parâmetro chave utilizado; reverse= true é para ordenar de forma decrescente
print (Lista)
```

```python
Lista=Lista0
Lista.sort(key = lambda item: item[1], reverse= True)     
print (Lista)
```

# Aula 19
Trocando o Valor entre Variáveis em Python

```python
x = 10
y = 'Pedro'

z = x
x = y
y = z

print(f'x={x} e y= {y}')
```

```python
x = 10
y = 'Pedro'

x, y = y, x
print(f'x={x} e y= {y}')
```
